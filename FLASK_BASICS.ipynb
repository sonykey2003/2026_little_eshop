## What is a web app?

When you visit a website, two computers talk:

```
   YOU (the browser)                    THE SERVER (Python)
  ┌─────────────────┐   "Can I have    ┌─────────────────┐
  │                 │    the page /?"  │                 │
  │   Chrome,       │ ───────────────► │   Your Flask    │
  │   Safari ...    │                  │   program       │
  │                 │ ◄─────────────── │                 │
  └─────────────────┘   "Here it is!"  └─────────────────┘
```

* The **browser** asks for a page.
* The **server** runs some Python and sends the page back.

**Flask** is a Python library that makes writing the server part easy. Today, **your laptop is the server.**

## Step 1: Hello, web! 

Here's a basic flask website Run the cell.

````python
from flask import Flask

app = Flask(__name__)

@app.route("/")
def home():
    return "Hello! This is my first website."

app.run(port=5001)
````

Now open **<http://127.0.0.1:5001>** in your browser. You should see your message!

### What does each line do?

| Code | Meaning |
| --- | --- |
| `app = Flask(__name__)` | Make a new web app. |
| `@app.route("/")` | *"When someone visits `/`, the home page…"* |
| `def home():` | *"…run this function."* |
| `return "Hello!..."` | Whatever you return gets sent to the browser. |
| `app.run(port=5001)` | Start the server and wait for visitors. |

> **127.0.0.1** means *"this computer"*. Only you can see your website right now.

## Step 2: More than one page 📄📄

Every page has its own **route** (address) and its own **function**.

Press **■ Stop** on the cell above first, then run this one.

````python
from flask import Flask

app = Flask(__name__)

@app.route("/")
def home():
    return "Welcome to the home page!"

@app.route("/about")
def about():
    return "This website was made by a Year 10 student."

app.run(port=5001)
````

## Step 3: Make it look like a real page 🎨

So far we've only sent plain text. Browsers understand **HTML**, the language of web pages. Tags like `<h1>` (a big heading), `<p>` (a paragraph) and `<a>` (a link) tell the browser what each piece is.

Press **■ Stop**, then run:

````python
from flask import Flask

app = Flask(__name__)

@app.route("/")
def home():
    return """
    <h1 style="color: teal;">Welcome to my shop!</h1>
    <p>We sell nice things at fair prices.</p>
    <a href="/about">About us</a>
    """

@app.route("/about")
def about():
    return """
    <h1>About us</h1>
    <p>We are a small shop run by students.</p>
    <a href="/">Back home</a>
    """

app.run(port=5001)
````

Open <http://127.0.0.1:5001> and click the links. The two pages now link to each other.

`style="color: teal;"` is a tiny bit of **CSS**, the language that controls how things *look*.

## Step 4: Pages that change 🔄

Here's where it gets clever. Part of the address can be a **variable**. Flask hands it to your function, so one function can make many different pages.

Press **■ Stop**, then run:

````python
from flask import Flask

app = Flask(__name__)

@app.route("/hello/<name>")
def hello(name):
    return f"<h1>Hello, {name}! 👋</h1>"

app.run(port=5001)
````

Try these. Put your own name in!

* <http://127.0.0.1:5001/hello/Sam>
* <http://127.0.0.1:5001/hello/Ms Lee>

`<name>` in the route catches whatever is typed there, and `f"...{name}..."` puts it into the page.

> 💡 This is how real websites work. A shop doesn't write a separate page for every product. It has **one** function for `/product/<number>`, and the number decides which product to show.

### ✏️ Try it

## Step 5: A mini shop 🛍️

Now let's show some **data**: a list of products stored in Python.

Typing HTML for every product by hand would be slow. Instead we write a **template**: HTML with a few special marks that Flask fills in.

| Template mark | Meaning |
| --- | --- |
| `{{ something }}` | Put a value here. |
| `{% for p in products %} ... {% endfor %}` | Repeat this part once for each product. |

Press **■ Stop**, then run:

````python
from flask import Flask, render_template_string

app = Flask(__name__)

# Our "database" for now: just a Python list.
products = [
    {"name": "Sun Hat",      "price": 390},
    {"name": "Tote Bag",     "price": 320},
    {"name": "Water Bottle", "price": 450},
]

PAGE = """
<h1>My Mini Shop</h1>
<ul>
  {% for p in products %}
    <li>{{ p.name }}: {{ p.price }} baht</li>
  {% endfor %}
</ul>
"""

@app.route("/")
def home():
    return render_template_string(PAGE, products=products)

app.run(port=5001)
````

Open <http://127.0.0.1:5001>. Three products, but we only wrote **one** `<li>` line. The loop made the others.

## Step 6: Let visitors add things (forms) ✍️

Until now, visitors could only **look**. A **form** lets them **send** something back to the server.

There are two kinds of request:

| | Means | Example |
| --- | --- | --- |
| **GET** | *"Show me a page."* | Typing an address, clicking a link |
| **POST** | *"Here's some information."* | Pressing a form's button |

Press **■ Stop**, then run:

````python
from flask import Flask, render_template_string, request, redirect

app = Flask(__name__)

products = [
    {"name": "Sun Hat",  "price": 390},
    {"name": "Tote Bag", "price": 320},
]

PAGE = """
<h1>My Mini Shop</h1>
<ul>
  {% for p in products %}
    <li>{{ p.name }}: {{ p.price }} baht</li>
  {% endfor %}
</ul>

<h2>Add a product</h2>
<form method="post">
  <input name="name"  placeholder="Product name">
  <input name="price" placeholder="Price">
  <button>Add</button>
</form>
"""

@app.route("/", methods=["GET", "POST"])
def home():
    if request.method == "POST":
        # Someone pressed the button. Read what they typed...
        new_product = {"name": request.form["name"], "price": request.form["price"]}
        products.append(new_product)
        # ...then send them back to the home page to see it.
        return redirect("/")

    return render_template_string(PAGE, products=products)

app.run(port=5001)
````

Open <http://127.0.0.1:5001>, add a product, and watch it appear in the list.

`request.form["name"]` reads whatever was typed in the box with `name="name"`.
